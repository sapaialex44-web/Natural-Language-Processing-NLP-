```python
import spacy
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
feedback_data = [
    {
        "feedback": "The service was excellent and the staff were friendly.",
        "contact_method": "email"
    },
    {
        "feedback": "I am unhappy with the long waiting time.",
        "contact_method": "chat"
    },
    {
        "feedback": "The product quality is amazing.",
        "contact_method": "email"
    },
    {
        "feedback": "The support team did not solve my problem.",
        "contact_method": "chat"
    },
    {
        "feedback": "I really enjoyed the experience.",
        "contact_method": "email"
    }
]

df = pd.DataFrame(feedback_data)

df
```

```python
def analyze_sentiment(text):
  blob=TextBlob(text)
  return blob.sentiment.polarity
```

```python
text="the service ws exellent"
score=analyze_sentiment(text)
print(score)
```

```python
def analyze_sentiment(text):
  blob=TextBlob(text)
  return blob.sentiment.polarity
  text="the service ws exellent"
  score = analyze_sentiment(text)
  print(score)
```

```python
df["sentiment_polarity"]=df["feedback"].apply(analyze_sentiment)
df
```

```python
def extract_entities(text):
  doc = nlp(text)
  entities = []
  for ent in doc.ents:
    entities.append(ent.text)
  return entities
```

```python
df["sentiment_polarity"]=df["feedback"].apply(extract_entities)
df
```

```python
df.to_csv("customer_feedback_results.csv", index=False)
```

```python
def sentiment_label(score):
  if score > 0:
    return "positive"
  elif score < 0:
    return "negative"
  else:
    return "neutral"
```

```python
df["sentiment_polarity"]=df["feedback"].apply(analyze_sentiment)
df["sentiment"]=df["sentiment_polarity"].apply(sentiment_label)
df
```
